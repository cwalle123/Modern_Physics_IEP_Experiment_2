**Method:** Monochromator setup: Hg or Na lamp -> slit -> collimating lens f1 -> 600 lines/mm diffraction grating (mounted on a rotating turntable with an angle scale) -> lens f2 -> screen (replaced by a camera for the doublet-splitting measurements). Rotating the grating brings each spectral line, in turn, onto a fixed point on the screen; the angle scale reading at that moment is converted to a wavelength.  
**Theory:** Diffraction grating equation d sin(u_m) = d sin(i) + m*lambda, with i = alpha + phi and u = alpha - phi (alpha = angle between the grating normal and the incident beam, phi = angle between the current grating orientation and the 0th-order/"mirror" position). Substituting gives 2 cos(alpha) sin(phi) = -mN*lambda (eq. II8_Que8), with N the number of grating lines per unit length. For unresolved doublets: Delta_lambda = cos(u)/(mN) * Delta_u, and resolving power R = lambda/Delta_lambda, with theoretical limit R_theor = m*N_tot.  
**Independent variable:** Diffraction order m (and which spectral line/colour is being measured).  
**Dependent variable:** Angle-scale reading at which each spectral line lands on the fixed point, for each order -- from this, phi and then lambda are calculated.  
**Controlled variablen:** Grating line density N (600 lines/mm); position of the fixed point on the screen/camera; alpha (fixed once the grating is aligned for a given lamp); slit width (kept constant during a given sweep); which lamp is in use (Hg or Na, one at a time).  
**Measurement instruments \& Settings:** Angle scale on the grating turntable (note its reading resolution/uncertainty); screen, for the direct angle-sweep measurements; camera, for the doublet-splitting measurements (note pixel size and the focal length of lens f2, needed to convert a pixel separation into Delta_u); adjustable slit (note width used).  
**Procedure:** (1) Build and collimate the Hg monochromator setup. (2) Autocollimate to i=0 and note that reading; pick a fixed screen point, rotate the grating until the 0th order lands on it, and read alpha and its uncertainty. (3) Sweep the grating through every visible order, logging colour/order/angle reading each time a line lands on the fixed point. (4) Compute phi, lambda and u(lambda) in Python; plot lambda vs order with literature dotted lines; take the weighted average per colour and compare to literature. (5) Swap in the camera, image the yellow Hg doublet, measure the pixel separation, and compute Delta_u, Delta_lambda and u(Delta_lambda). (6) Repeat steps 2-5 for the Na lamp, and also attempt the splitting measurement for every Na line. (7) Evaluation: compute the Na doublet separation and resolving power (measured vs theoretical) at every order measured.
